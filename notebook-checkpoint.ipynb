```python
from game import Game
import time
import os
import re
from collections import deque
import copy
import heapq
```

## Load maps

```python
def extract_map_files(directory):
    pattern = re.compile(r'^map(\d+)\.txt$')
    map_file_indices = []

    for file_name in os.listdir(directory):
        match = pattern.match(file_name)
        if match:
            map_file_indices.append(match.group(1))

    return [int(idx) for idx in map_file_indices]

def is_valid_input(map, indices, algorithm, solvers):
    valid_input = True
    if map not in indices:
        print(f"Map index out of range. Please choose within range {min(indices)} to {max(indices)}")
        valid_input = False
    if algorithm not in solvers.keys():    
        print(f"{algorithm} is not a defined algorithm. Please choose from", *[f"{solver} ({i+1})  " for i, solver in enumerate(solvers.keys())])
        valid_input = False
    return valid_input

def load_map(map_index):  
    file_name = "map" + str(map_index) + ".txt"
    with open('./assets/maps/' + file_name) as f:
        game_map = f.read()
    return game_map

map_file_indices = extract_map_files("./assets/maps/")
```

## Solvers

### BFS

```python
# TODO: Must return moves (if there is no solution return None), number of visited states


def solver_bfs(game_map):
    game = Game(game_map)
    initial_state = (game.get_player_position(), tuple(game.get_box_locations()))
    queue = deque([(initial_state, "")])
    visited = set()
    visited.add(initial_state)
    while queue:
        (player_pos, boxes), path = queue.popleft()
        game.set_player_position(player_pos)
        game.set_box_positions(list(boxes))
        if game.is_game_won():
            return path, len(visited)
        
        for move in "UDLR":
            prev_player_pos = game.get_player_position()
            prev_boxes = list(game.get_box_locations())
            if game.apply_move(move):
                new_state = (game.get_player_position(), tuple(game.get_box_locations()))
                if new_state not in visited:
                    visited.add(new_state)
                    queue.append((new_state, path + move))
            game.set_player_position(prev_player_pos)
            game.set_box_positions(prev_boxes)
    return None, len(visited)
map = load_map(5)
solver_bfs(map)
```

### DFS

```python
# TODO: Must return moves, number of visited states
from collections import deque

def solver_dfs(game_map):
    game = Game(game_map)
    max_depth = 1000;
    initial_state = (game.get_player_position(), tuple(game.get_box_locations()))
    stack = deque([(initial_state, "")])
    visited = set()
    visited.add(initial_state)
    while stack:
        (player_pos, boxes), path = stack.pop()
        game.set_player_position(player_pos)
        game.set_box_positions(list(boxes))
        
        if len(path) >= max_depth:
            continue
        if game.is_game_won():
            return path, len(visited)
            
        for move in "UDLR":
            prev_player_pos = game.get_player_position()
            prev_boxes = list(game.get_box_locations())

            if game.apply_move(move):
                new_state = (game.get_player_position(), tuple(game.get_box_locations()))

                if new_state not in visited:
                    visited.add(new_state)
                    stack.append((new_state, path + move))
            game.set_player_position(prev_player_pos)
            game.set_box_positions(prev_boxes)
    return None, 0
map = load_map(10)
solver_dfs(map)
```

### IDS

```python
# TODO: Must return moves, number of visited states

def dls(game, state, depth, path, visited, moves_order="UDLR"):

    game.set_player_position(state[0])
    game.set_box_positions(list(state[1]))
    
    if game.is_game_won():
        return path
    
    if depth == 0:
        return None
    
    for move in moves_order:
        prev_player_pos = game.get_player_position()
        prev_boxes = list(game.get_box_locations())
        

        if game.apply_move(move):
            new_state = (game.get_player_position(), tuple(game.get_box_locations()))
            if new_state not in visited:
                visited.add(new_state)
                result = dls(game, new_state, depth - 1, path + move, visited, moves_order)
                if result is not None:
                    return result
                visited.remove(new_state)
        
        game.set_player_position(prev_player_pos)
        game.set_box_positions(prev_boxes)
    
    return None



def solver_ids(game_map):
    game = Game(game_map)
    max_depth=50
    initial_state = (game.get_player_position(), tuple(game.get_box_locations()))
    for depth in range(max_depth + 1):
        visited = set([initial_state])
        result = dls(game, initial_state, depth, "", visited, "UDLR")
        if result is not None:
            return result, len(visited)
    return None, 0
map = load_map(8)
solver_ids(map)
```

### A*

```python
# TODO

# TODO: Must return moves, number of visited states

def manhattan_distance(pos1, pos2):
    return abs(pos1[0] - pos2[0]) + abs(pos1[1] - pos2[1])

def heuristic(state, game):
    (player_pos, boxes) = state
    total_dist = 0
    for i, box_pos in enumerate(boxes):
        goal_pos = game.goals[i]
        total_dist += manhattan_distance(box_pos, goal_pos)
    return total_dist

def solver_astar(game_map, heuristic_func=heuristic, w=1.5):
    game = Game(game_map)
    initial_state = (game.get_player_position(), tuple(game.get_box_locations()))
    
    initial_h = heuristic(initial_state, game)
    frontier = []
    heapq.heappush(frontier, (initial_h * w, 0, initial_state, "")) 
    cost_so_far = {initial_state: 0}
    
    while frontier:
        current_f, current_g, current_state, current_path = heapq.heappop(frontier)
        (player_pos, boxes) = current_state
        
        game.set_player_position(player_pos)
        game.set_box_positions(list(boxes))
        
        if game.is_game_won():
            return current_path, len(cost_so_far)
        for move in "UDLR":
            old_player_pos = game.get_player_position()
            old_boxes = list(game.get_box_locations())
            
            if game.apply_move(move):
                new_state = (game.get_player_position(), tuple(game.get_box_locations()))
                new_g = current_g + 1  
                
                if new_state not in cost_so_far or new_g < cost_so_far[new_state]:
                    cost_so_far[new_state] = new_g
                    new_h = heuristic(new_state, game)
                    new_f = new_g + w * new_h
                    heapq.heappush(frontier, (new_f, new_g, new_state, current_path + move))
            
            game.set_player_position(old_player_pos)
            game.set_box_positions(old_boxes)
    return None, 0
map = load_map(10)
solver_astar(map)
```

## Solve

```python
SOLVERS = {
    "BFS": solver_bfs,
    "DFS": solver_dfs,
    "IDS": solver_ids,
    "A*": solver_astar
}
```

```python
def solve(map, method):  
    
    if not is_valid_input(map, map_file_indices, method, SOLVERS):
        return
    
    file_name = "map" + str(map) + ".txt"
    with open('./assets/maps/' + file_name) as f:
        game_map = f.read()
    
    start_time = time.time()
    moves, numof_visited_states = SOLVERS[method](game_map)
    end_time = time.time()
    print(f"{method} took {round(end_time - start_time, 2)} seconds on map {map} and visited {numof_visited_states} states.")
    
    if moves is None:
        print("No Solution Found!")
    else:
        print(f"{len(moves)} moves were used: {moves}")
            
```

```python
solve(1, "BFS") # Solve map 1 using BFS
```

```python
def solve_all():
    for map in range(min(map_file_indices), max(map_file_indices) + 1):
        for method in SOLVERS.keys():
            if ((method == "DFS" and (map == 7 or map == 9)) or (method == "IDS" and (map == 6 or map == 7 or map == 9 or map == 10))):
                print("No answer")
                continue
            if ((method == "BFS" or method == "A*") and map == 9):
                print("No answer")
                continue
            solve(map, method)
            
```
